```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
import ast

def convert_genres(genres):
    genres = ast.literal_eval(genres)
    return [genre['name'].lower() for genre in genres]
```

```python
movies['genres'] = movies['genres'].apply(convert_genres)
```

```python
movies.head(2)
```

```python
movies['keywords'] = movies['keywords'].apply(convert_genres)
```

```python
def convert_cast(cast):
    cast = ast.literal_eval(cast)
    return [actor['name'].lower() for actor in cast[:15]]

movies['cast'] = movies['cast'].apply(convert_cast)
```

```python
movies.head(2)
```

```python
movies.iloc[0]['cast']
```

```python
 # "job": "Director", "name": "James Cameron"
```

```python
def get_Director_name(crew):
    l=[]
    director = ast.literal_eval(crew)
    for person in director:
        if person['job'] == "Director":
            l.append(person['name'].lower())

    return l      
```

```python
movies['crew']= movies['crew'].apply(get_Director_name)
```

```python
movies.head(2)
```

```python
#convert string into list 
movies['overview']= movies['overview'].apply(lambda x: x.split())
```

```python
movies.head(2)
```

```python
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(' ' ,'') for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(' ' ,'') for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(' ' ,'') for i in x])
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(' ' ,'') for i in x])
```

```python
movies.head(2)
```

```python
movies['tag']= movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```
